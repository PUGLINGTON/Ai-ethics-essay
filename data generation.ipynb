```python
#non random evenly spaced variables

import numpy as np
import pandas as pd

# Functions
def linear(x):
    return 2 * x + 5

def polynomial(x):
    return x**3 - 4*x**2 + x + 6

def challenge(x):
    return np.sin(x) * np.exp(x / 5)

# Generate 1000 x values between -20 and 20 (1 decimal place)
x = np.round(np.linspace(-20, 20, 1000), 1)

# Build DataFrame
df = pd.DataFrame({
    'x':           x,
    'f_linear':    np.round(linear(x), 4),
    'g_polynomial': np.round(polynomial(x), 4),
    'h_challenge': np.round(challenge(x), 4)
})

# Save to CSV
df.to_csv('non-random-variable.csv', index=False)

print(df.head(10))
print(f"\nShape: {df.shape}")
```

```python
#random with duplicates

import numpy as np
import pandas as pd

# Functions
def linear(x):
    return 2 * x + 5

def polynomial(x):
    return x**3 - 4*x**2 + x + 6

def challenge(x):
    return np.sin(x) * np.exp(x / 5)

# Generate 1000 random x values between -20 and 20 (1 decimal place)
np.random.seed(42)
x = np.round(np.random.uniform(-20, 20, 1000), 1)
x = np.sort(x)  # optional: sort so the CSV reads sequentially

# Build DataFrame
df = pd.DataFrame({
    'x':            x,
    'f_linear':     np.round(linear(x), 4),
    'g_polynomial': np.round(polynomial(x), 4),
    'h_challenge':  np.round(challenge(x), 4)
})

# Save to CSV
df.to_csv('random-duplicate.csv', index=False)

print(df.head(10))
print(f"\nShape: {df.shape}")
```

```python
# unique values


import numpy as np
import pandas as pd

# Functions
def linear(x):
    return 2 * x + 5

def polynomial(x):
    return x**3 - 4*x**2 + x + 6

def challenge(x):
    return np.sin(x) * np.exp(x / 5)

# Generate all 401 unique x values, randomly shuffled
np.random.seed(42)
x = np.random.choice(np.arange(-20, 20.1, 0.1), size=401, replace=False)
x = np.round(x, 1)  # clean up any floating point drift from arange

# Build DataFrame
df = pd.DataFrame({
    'x':            x,
    'f_linear':     np.round(linear(x), 4),
    'g_polynomial': np.round(polynomial(x), 4),
    'h_challenge':  np.round(challenge(x), 4)
})

# Save to CSV
df.to_csv('unique-401.csv', index=False)

print(df.head(10))
print(f"\nShape: {df.shape}")  # (401, 4)

#df.sort_values("x")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --- Load the three datasets ---
df_unique    = pd.read_csv('unique-401.csv')
df_random    = pd.read_csv('random-duplicate.csv')
df_nonrandom = pd.read_csv('non-random-variable.csv')

datasets = {
    'unique-401':       df_unique,
    'random-duplicate': df_random,
    'non-random':       df_nonrandom,
}

# --- True curves for overlay ---
x_true = np.linspace(-20, 20, 500)
true_curves = {
    'f_linear':     2 * x_true + 5,
    'g_polynomial': x_true**3 - 4*x_true**2 + x_true + 6,
    'h_challenge':  np.sin(x_true) * np.exp(x_true / 5),
}

true_colors = {
    'f_linear':     '#378ADD',
    'g_polynomial': '#1D9E75',
    'h_challenge':  '#7F77DD',
}

labels = {
    'f_linear':     'f(x) = 2x + 5',
    'g_polynomial': 'g(x) = x³ − 4x² + x + 6',
    'h_challenge':  'h(x) = sin(x) · e^(x/5)',
}

# --- Plot: 3 rows (one per function) x 3 cols (one per dataset) ---
fig, axes = plt.subplots(3, 3, figsize=(16, 12))
fig.suptitle('Functions vs datasets', fontsize=14, fontweight='normal', y=1.01)

for row, col_name in enumerate(['f_linear', 'g_polynomial', 'h_challenge']):
    for col, (ds_name, df) in enumerate(datasets.items()):

        ax = axes[row][col]
        df_sorted = df.sort_values('x')

        # Scatter: sampled points
        ax.scatter(df_sorted['x'], df_sorted[col_name],
                   color='#D85A30', alpha=0.4, s=8, label='sampled points')

        # Line: true curve
        ax.plot(x_true, true_curves[col_name],
                color=true_colors[col_name], linewidth=1.8, label='true curve')

        # Labels
        if row == 0:
            ax.set_title(ds_name, fontsize=11, fontweight='normal')
        if col == 0:
            ax.set_ylabel(labels[col_name], fontsize=9)

        ax.set_xlabel('x', fontsize=8)
        ax.tick_params(labelsize=7)
        ax.grid(True, alpha=0.2)
        ax.spines[['top', 'right']].set_visible(False)

        if row == 0 and col == 0:
            ax.legend(fontsize=7)

plt.tight_layout()
plt.savefig('functions_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)

noise_scales = {
    'f_linear':     2.0,
    'g_polynomial': 500.0,
    'h_challenge':  0.5,
}

def add_noise(df):
    df = df.copy()
    for col, scale in noise_scales.items():
        df[col] = df[col] + np.random.normal(0, scale, size=len(df))
        df[col] = df[col].round(4)
    return df

files = {
    'unique-401.csv':       'unique-401-noisy.csv',
    'random-duplicate.csv': 'random-duplicate-noisy.csv',
    'non-random-variable.csv': 'non-random-noisy.csv',
}

for input_file, output_file in files.items():
    df = pd.read_csv(input_file)
    df_noisy = add_noise(df)
    df_noisy.to_csv(output_file, index=False)
    print(f'{input_file} → {output_file} done')
```
